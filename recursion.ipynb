```python
def p(i,n):
    if i>n:
        return
    print(i)
    p(i+1,n)
p(0,5)
```

```python
def p(n):
    if n==0:
        return
    print(n)
    p(n-1)
p(5)
```

```python
def func(n):
    if n==0:
        return 1
    return n*func(n-1)
print(func(5))
```

```python
def s(n):
    if n==0:
        return 0
    return n+s(n-1)
print(s(5))
```

```python
def addDigits(num):

    if num < 10:
        return num

    def func(n):

        if n == 0:
            return 0

        return n % 10 + func(n // 10)

    s = func(num)
    if s < 10:
        return s
    return addDigits(s)

addDigits(38)
```
